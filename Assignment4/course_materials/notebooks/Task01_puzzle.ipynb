```python
github_storage = "https://raw.githubusercontent.com/FacultadInformatica-LinkedData/Curso2026-2027-Datascience/refs/heads/master/Assignment4/course_materials"
```

```python
from rdflib import Graph, Namespace, Literal
g = Graph()
```

Spanish: Para leer y escribir ficheros que contienen tripletas en RDFlib empleamos dos métodos de la librería: parse y serialize. Podemos añadir tripletas a nuestro grafo empleando parse, que leerá el recurso proporcionado. Debemos además indicarle el formato si este no puede ser inferido. El recurso puede ser local o remoto, como en nuestro caso. El resultado es el mismo. Podemos añadir todos los datos que queramos a nuestro grafo, los datos simplemente se irán fusionando.

English: To read and write RDF files that contain triplets we employ two methods from the library: parse and serialize. We can add triplets to our graph using parse, which will read the specified resource. We must also specify the format if it cannot be inferred. The resource can be local or remote, as in our case. The result is the same. We can add as much data as we want to our graph, and the data will simple be merged together.

```python
g.parse(github_storage+"/rdf/example1.rdf", format="xml")
```

```python
g.parse(github_storage+"/rdf/example2.rdf", format="xml")
```

```python
g.serialize("example1.ttl", format="ttl")
```

Spanish:  Ahora podemos comprobar el resultado volcando las tripletas de forma sencilla.

English: We can now check the result by simply outputting the triplets.

```python
for subj, pred, obj in g:
  print(subj,pred,obj)
```

Spanish: Ahora podemos ver la operación inversa, serializando estos datos a algún formato que nos lo permita. Este proceso también nos permite una conversión sencilla entre formatos. Hacer primero en ttl y luego en xml.

English: We can now look at the reverse process: serialising this data into a format that allows us to do so. This process also enables us to convert easily between formats. First in TTL, then in XML.

```python
print(g.serialize(format="ttl"))
```

```python
print(g.serialize(format="xml"))
```

Spanish: También podemos guardar el resultado serializado en un fichero, puedes ver este fichero resultante en el panel izquierdo.

English: We can also save the serialised result to a file; you can view this resulting file in the left-hand pane.
